| Encoder            | Best suited for                                          | Main limitation                                                               |
| ------------------ | -------------------------------------------------------- | ----------------------------------------------------------------------------- |
| LabelEncoder       | Target class labels                                      | Can create misleading numerical order if used carelessly for nominal features |
| OrdinalEncoder     | Categories with a meaningful order                       | Numeric gaps do not necessarily represent real distances                      |
| OneHotEncoder      | Nominal features with manageable cardinality             | Can create many columns                                                       |
| `pd.get_dummies()` | Convenient DataFrame transformations                     | Independently transformed datasets may have mismatched columns                |
| BinaryEncoder      | High-cardinality nominal features                        | Binary digits are not inherently meaningful or ordered                        |
| Frequency Encoding | High-cardinality features where frequency is informative | Categories with equal frequency become indistinguishable                      |
| TargetEncoder      | Categories whose target statistics may be predictive     | Target leakage and overfitting risk                                           |

# 2. Label Encoding

## Definition

Label Encoding assigns an integer to each distinct category in a categorical variable.

For example:

| Department | Encoded Value |
|---|---:|
| AIML | 0 |
| CSE | 1 |
| ECE | 2 |

The encoder creates a mapping between categories and integer values.

## How Does It Work?

Step 1: Identify the distinct categories.

Step 2: Assign an integer to each category.

Step 3: Replace each original category with its assigned integer.

Step 4: Apply the same mapping when transforming new data.


```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

# fit() learns the distinct categories and their mapping.
# transform() converts the categories into integers

df["Department_Label"] = label_encoder.fit_transform(
    df["Department"]
)

print(df[["Department", "Department_Label"]])
```

```python
label_encoder.transform(label_encoder.classes_)
```

```python
print("Classes:", label_encoder.classes_)  # classes_ stores the categories recognized by the encoder

for category, encoded_value in zip(
    label_encoder.classes_,
    label_encoder.transform(label_encoder.classes_)
):
    print(category, "->", encoded_value)
```

```python
encoded_values = label_encoder.transform(
    df["Department"]
)

decoded_values = label_encoder.inverse_transform(
    encoded_values
)

print("Encoded:", encoded_values)
print("Decoded:", decoded_values)
```

# 3. Ordinal Encoding

## Definition

Ordinal Encoding converts categorical values into numerical values according to a meaningful order.

Example:

| Skill Level | Encoded Value |
|---|---:|
| Beginner | 0 |
| Intermediate | 1 |
| Advanced | 2 |

Unlike arbitrary integer assignment, the ordering is explicitly defined.

## Why Is It Useful?

Suppose a student has one of three skill levels:

Beginner < Intermediate < Advanced

These categories have a meaningful ranking.

Ordinal Encoding preserves this ranking in numerical form.

## Important Limitation

Ordinal Encoding introduces an ordered numerical representation. It does not guarantee that the distance between adjacent categories has a meaningful numerical interpretation.

For example:

Beginner = 0
Intermediate = 1
Advanced = 2

This representation does not prove that the difference between Beginner and Intermediate is exactly equal to the difference between Intermediate and Advanced.



## **OrdinalEncoder expects a two-dimensional feature matrix, so we use the double-bracket form.**

```python
from sklearn.preprocessing import OrdinalEncoder

ordinal_encoder = OrdinalEncoder(
    categories=[["Beginner", "Intermediate", "Advanced"]]
)

# OrdinalEncoder expects a two-dimensional feature matrix, so we use the double-bracket form.

df["Skill_Ordinal"] = ordinal_encoder.fit_transform(
    df[["Skill_Level"]]
)

print(df[["Skill_Level", "Skill_Ordinal"]])
```

```python
education_encoder = OrdinalEncoder(
    categories=[["Bachelor", "Master", "PhD"]]
)

df["Education_Ordinal"] = education_encoder.fit_transform(
    df[["Education_Level"]]
)

print(df[["Education_Level", "Education_Ordinal"]])
```

# 4. One-Hot Encoding

## Definition

One-Hot Encoding converts each distinct category into a separate binary column.

Each original category is represented by:
- 1 if the row belongs to that category.
- 0 otherwise.

## Example

Original feature:

| City |
|---|
| Hyderabad |
| Chennai |
| Vijayawada |
| Hyderabad |

After One-Hot Encoding:

| City_Chennai | City_Hyderabad | City_Vijayawada |
|---:|---:|---:|
| 0 | 1 | 0 |
| 1 | 0 | 0 |
| 0 | 0 | 1 |
| 0 | 1 | 0 |

## How Does It Work?

Step 1: Identify all categories.

Step 2: Create one binary column for each category.

Step 3: For each row, place 1 in the matching category column.

Step 4: Place 0 in the other category columns.

```python
from sklearn.preprocessing import OneHotEncoder

onehot_encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown="ignore"
)

city_encoded = onehot_encoder.fit_transform(
    df[["City"]]
)

city_columns = onehot_encoder.get_feature_names_out(["City"]) # This method is used to name the out cols with city 

city_encoded_df = pd.DataFrame(
    city_encoded,
    columns=city_columns,
    index=df.index
)

print(city_encoded_df)
```

```python
nominal_columns = ["Department", "Gender", "City"]

onehot_encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown="ignore"
)

encoded_array = onehot_encoder.fit_transform(
    df[nominal_columns]
)

encoded_columns = onehot_encoder.get_feature_names_out(
    nominal_columns
)

encoded_df = pd.DataFrame(
    encoded_array,
    columns=encoded_columns,
    index=df.index
)

print(encoded_df.head())
```

# 5. Pandas get_dummies()

## Definition

Pandas provides the get_dummies() function to convert categorical variables into indicator columns.

It is a convenient way to perform One-Hot Encoding using Pandas.

## How Does It Work?

Step 1: Identify categorical columns.

Step 2: Identify the unique categories in each selected column.

Step 3: Create a separate indicator column for each category.

Step 4: Assign 1 when the category matches and 0 otherwise.

## Example

Original:

| Department |
|---|
| AIML |
| CSE |
| ECE |

Encoded:

| Department_AIML | Department_CSE | Department_ECE |
|---:|---:|---:|
| 1 | 0 | 0 |
| 0 | 1 | 0 |

## Difference from Scikit-learn OneHotEncoder

Pandas get_dummies() is convenient for quick DataFrame transformations.

Scikit-learn OneHotEncoder is often preferable in a machine learning pipeline because it learns a consistent category mapping during fitting and can handle unseen categories according to its configuration.

For production machine learning workflows, fit the encoder on training data and reuse the same fitted encoder for validation, testing, and new data.

# 6. Binary Encoding

## Definition

Binary Encoding converts categories into integer identifiers and then represents those identifiers using binary digits.

For example, suppose a feature has eight distinct categories.

A One-Hot Encoder may create eight columns.

Binary Encoding can represent eight distinct identifiers using three binary digits because:

2^3 = 8

## How Does It Work?

Step 1: Assign an integer identifier to each category.

Step 2: Convert each identifier into binary.

Step 3: Represent each binary digit in a separate column.

For example:

| Category ID | Binary Representation |
|---:|---|
| 0 | 000 |
| 1 | 001 |
| 2 | 010 |
| 3 | 011 |
| 4 | 100 |


## Appropriate Use Cases

Binary Encoding can be useful for high-cardinality nominal features when the number of One-Hot columns would become large.

It is not automatically better than One-Hot Encoding for every dataset.

## **Binary Encoding is not included in the standard Scikit-learn preprocessing encoders. One common implementation is available through the category_encoders package**

# 7. Frequency Encoding

## Definition

Frequency Encoding replaces each category with the number of times it appears in the dataset, or with its relative frequency.

Example:

| City |
|---|
| Hyderabad |
| Chennai |
| Hyderabad |
| Vijayawada |
| Hyderabad |
| Chennai |

Category counts:

- Hyderabad: 3
- Chennai: 2
- Vijayawada: 1

Frequency Encoding produces:

| City | Frequency |
|---|---:|
| Hyderabad | 3 |
| Chennai | 2 |
| Hyderabad | 3 |
| Vijayawada | 1 |
| Hyderabad | 3 |
| Chennai | 2 |

## How Does It Work?

Step 1: Count how often each category occurs.

Step 2: Build a mapping from category to count or relative frequency.

Step 3: Replace each category with its mapped value.


```python
city_frequency = df["City"].value_counts()

print(city_frequency)
```

```python
frequency_mapping = df["City"].value_counts().to_dict()
# .to_dict() converts series into a python dict 

print(frequency_mapping)
```

### **In a real predictive workflow, learn the category-frequency mapping from the training set and reuse it on validation and test data**

```python
df["City_Frequency"] = df["City"].map(
    frequency_mapping
)

print(df[["City", "City_Frequency"]])
```

# 8. Target Encoding

## Definition

Target Encoding replaces a category with a statistic calculated from the target values associated with that category.

For binary classification, a common approach is to replace each category with the mean of the target values for samples in that category.

Suppose the target is:

Placed = 1
Not Placed = 0

Imagine the following data:

| Department | Placement |
|---|---:|
| AIML | 1 |
| AIML | 1 |
| AIML | 0 |
| CSE | 1 |
| CSE | 0 |
| ECE | 0 |

For AIML:

Mean target = (1 + 1 + 0) / 3 = 0.667

For CSE:

Mean target = (1 + 0) / 2 = 0.5

For ECE:

Mean target = 0 / 1 = 0

The encoded values are therefore:

AIML = 0.667
CSE = 0.500
ECE = 0.000

## How Does It Work?

Step 1: Group samples by category.

Step 2: Calculate a target statistic for each category.

Step 3: Build a mapping from category to target statistic.

Step 4: Replace the categories with the corresponding statistic.

## Important Rule

Target Encoding must be implemented without allowing each training sample's target to leak directly into its own encoded feature.

Cross-fitting, out-of-fold encoding, and smoothing are common ways to reduce this risk.

```python
df["Placed_Target"] = (df["Placement_Status"] == "Placed").astype(int)   # (True.)astype(int) is a number 1 

print(
    df[[
        "Department",
        "Placement_Status",
        "Placed_Target"
    ]]
)
```

##### **Mean target for each one**

```python
department_target_mean = df.groupby("Department")["Placed_Target"].mean()

print(department_target_mean)
```

#### **Apply the target Mapping**

```python
df["Department_Target_Encoded"] = df["Department"].map(department_target_mean) # MAP_THIS_USING.map(...)

print(
    df[[
        "Department",
        "Placed_Target",
        "Department_Target_Encoded"
    ]]
)
```

```python
from sklearn.preprocessing import TargetEncoder
```

## Scikit-learn TargetEncoder

Scikit-learn's TargetEncoder converts categories using target statistics.

For binary classification, the encoded value is based on the target mean.

Unlike a simple manually calculated mapping, TargetEncoder supports cross-fitting during fit_transform().

Cross-fitting helps prevent each training sample's target from being directly revealed through its own encoded feature.

The encoder also uses smoothing to help control the effect of categories with few observations.

| Parameter              | Meaning                                                     |
| ---------------------- | ----------------------------------------------------------- |
| `target_type="binary"` | Specifies binary classification                             |
| `smooth="auto"`        | Determines smoothing automatically                          |
| `cv=5`                 | Uses five-fold cross-fitting when encoding training samples |
| `random_state=42`      | Makes randomized cross-fitting reproducible                 |


```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import TargetEncoder

X = df[["Department"]]
y = df["Placed_Target"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

target_encoder = TargetEncoder(
    target_type="binary",
    smooth="auto",
    cv=5,
    random_state=42
)

X_train_encoded = target_encoder.fit_transform(
    X_train,
    y_train
)

X_test_encoded = target_encoder.transform(
    X_test
)

print("Training encodings:")
print(X_train_encoded)

print("\nTesting encodings:")
print(X_test_encoded)
```

# 9. Comparing Encoding Techniques

Different encoders represent the same categorical feature in different ways.

For the Department feature, the distinct categories are:

- AIML
- CSE
- ECE

The encoders have different goals:

- Label Encoding: Assigns integer identifiers.
- Ordinal Encoding: Assigns numbers according to an explicit order.
- One-Hot Encoding: Creates binary indicator columns.
- Binary Encoding: Represents category identifiers using binary digits.
- Frequency Encoding: Represents categories using their frequencies.
- Target Encoding: Represents categories using target statistics.

No single encoding technique is best for every dataset and algorithm.

```python
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder, OneHotEncoder

department = df[["Department"]]

label_encoder = LabelEncoder()
label_result = label_encoder.fit_transform(
    department["Department"]
)

ordinal_encoder = OrdinalEncoder(
    categories=[["AIML", "CSE", "ECE"]]
)
ordinal_result = ordinal_encoder.fit_transform(
    department
)

onehot_encoder = OneHotEncoder(
    sparse_output=False
)
onehot_result = onehot_encoder.fit_transform(
    department
)

print("Label Encoding:")
print(label_result[:5])

print("\nOrdinal Encoding:")
print(ordinal_result[:5])

print("\nOne-Hot Encoding:")
print(onehot_result[:5])
```

# 10. Data Leakage in Categorical Encoding

## What Is Data Leakage?

Data leakage occurs when information that would not legitimately be available during training influences the training process.

For categorical encoding, leakage can happen when an encoder learns information from validation or test data that should have remained unseen.

## Incorrect Workflow

1. Load the complete dataset.
2. Calculate category statistics using all rows.
3. Split the data into training and testing sets.
4. Train and evaluate the model.

This can leak information, particularly when encodings use target statistics.

## Correct General Workflow

1. Split the dataset into training and testing sets.
2. Fit the encoder using training data only.
3. Transform the training data appropriately.
4. Transform the test data using the fitted encoder.
5. Train the model on the encoded training data.
6. Evaluate it on the encoded test data.

## Special Rule for Target Encoding

Target Encoding needs extra care because the encoded values depend on target labels.

For training rows, use cross-fitting or another leakage-safe method.

For validation and test rows, use statistics learned exclusively from the training data.
