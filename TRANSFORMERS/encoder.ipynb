| Order | Component                          | Purpose                                                |
| ----- | ---------------------------------- | ------------------------------------------------------ |
| 1     | Input embeddings                   | Convert token IDs into vectors                         |
| 2     | Positional encoding                | Give the model information about token positions       |
| 3     | Scaled dot-product attention       | Calculate attention outputs from Q, K, and V           |
| 4     | Self-attention                     | Let tokens attend to other tokens in the same sequence |
| 5     | Padding and causal masks           | Control which tokens can be attended to                |
| 6     | Multi-head attention               | Run several attention operations in parallel           |
| 7     | Feed-forward network               | Transform each token's representation                  |
| 8     | Residual connections and LayerNorm | Help stabilize and train deeper networks               |
| 9     | Transformer block                  | Combine the components into one block                  |
