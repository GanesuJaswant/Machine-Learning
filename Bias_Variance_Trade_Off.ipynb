| Property                   | Underfitting | Good fit          | Overfitting                   |
| -------------------------- | ------------ | ----------------- | ----------------------------- |
| Bias                       | High         | Appropriate       | Often low                     |
| Variance                   | Often low    | Appropriate       | High                          |
| Model complexity           | Too low      | Suitable          | Often too high                |
| Training performance       | Poor         | Good              | Excellent                     |
| Testing performance        | Poor         | Good              | Often poor                    |
| Learns important patterns? | Not enough   | Yes               | Yes, but may also learn noise |
| Main problem               | High bias    | Neither dominates | High variance                 |


# **BIAS**

#### **Bias** is the error caused by a model making overly simple assumptions about the data.
#### In simple words, the model is unable to learn the actual relationship between inputs and outputs.
High bias means the model is not learning enough.

1. Characteristics of high bias
2. The model is too simple.
3. It fails to learn important patterns.
4. It performs poorly on training data.
5. It usually performs poorly on testing data.
6. It is associated with underfitting.

# **VARIANCE**

#### **Variance** describes how sensitive a model is to changes in its training data.
#### In simple words, the model learns the training data too closely, so small changes in the training dataset can produce substantially different models.

1. haracteristics of high variance
2. The model is excessively complex for the available data.
3. It may learn noise in the training dataset.
4. It performs very well on training data.
5. It performs considerably worse on unseen data.
6. It is associated with overfitting
